# TeraScan basics

## Connect

```python
import time
from DLCsmartTHz_V3_0_0 import DLCsmartTHz, NetworkConnection

dlc = DLCsmartTHz(NetworkConnection('1.1.1.2', timeout=15))
dlc.open()
print('connected')
```

## Firmware and serial

```python
print('firmware', dlc.general.fw_ver.get())
print('serial  ', dlc.general.serial_number.get())
```

## Emission on

```python
dlc.laser_operation.laser1.emission_enable.set(True)
dlc.laser_operation.laser2.emission_enable.set(True)
dlc.laser_operation.emission_global_enable.set(True)
print('lasers enabled')
```

## Is it emitting

```python
print('emission', dlc.laser_operation.emission.get())
```

## Set the TX frequency

GHz. The sleep is the tuning time.

```python
dlc.frequency.frequency_set.set(650.0)
time.sleep(0.5)
print('set to', dlc.frequency.frequency_set.get(), 'GHz')
```

## Read the actual frequency

```python
print('actual', dlc.frequency.frequency_act.get(), 'GHz')
```

## Set the chopping frequency

```python
dlc.lockin.frequency.set(1000.0) #Chopping frequency is in Hz
print('chop', dlc.lockin.frequency.get(), 'Hz')
```

## Set the integration time

The controller quantises this, so read it back.

```python
dlc.lockin.integration_time.set(0.080)
print('integration', dlc.lockin.integration_time.get())
```

## Read the lock-in

Returns `(value, valid)`.

```python
value, valid = dlc.lockin.lock_in_value_nanoamp.get()
print(value, 'nA  valid:', valid)
```

## Step through a range

```python
for f in [600.0, 601.0, 602.0, 603.0, 604.0]:
    dlc.frequency.frequency_set.set(f)
    time.sleep(0.5)
    print(dlc.frequency.frequency_act.get(), dlc.lockin.lock_in_value_nanoamp.get())
```

## Read the network settings

```python
print('ip     ', dlc.net_conf.ip_addr.get())
print('netmask', dlc.net_conf.net_mask.get())
print('dhcp   ', dlc.net_conf.dhcp.get())
```

## Set the IP address

`apply()` drops this connection; reconnect at the new address.

```python
dlc.net_conf.set_ip('192.168.1.50', '255.255.255.0')
dlc.net_conf.apply()
print('applied, connection dropped')
```

## Switch to DHCP

```python
dlc.net_conf.set_dhcp()
dlc.net_conf.apply()
print('applied, connection dropped')
```

## Emission off

```python
dlc.laser_operation.emission_global_enable.set(False)
print('emission', dlc.laser_operation.emission.get())
```

## Close

```python
dlc.close()
print('closed')
```
